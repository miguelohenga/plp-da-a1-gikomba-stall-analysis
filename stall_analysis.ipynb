## Task 1:Basic day summary
*My approach:I will use built-in Python functions like 'sum()', 'len()','max()' and 'min()'to calculate the overal stats.Then I will use 'list.index()' to find the positions of the best and worst days.*

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
        2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {

  "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},

  "Onions":  {"qty": 40, "reorder": 20, "cost": 70, "price": 100},

  "Sukuma":  {"qty": 9, "reorder": 15, "cost": 30, "price": 50},

  "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},

  "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},

}

#1.Calculate totals
total_sales=sum(daily_sales)
average_sales=total_sales/len(daily_sales)
highest_sale=max(daily_sales)
lowest_sale=min(daily_sales)

#2.Find the index of the highest and lowest sales
best_day_index=daily_sales.index(highest_sale)
worst_day_index=daily_sales.index(lowest_sale)

#3.The day name and week number
best_day_name=days[best_day_index]
best_week=1 if best_day_index < 7 else 2

worst_day_name=days[worst_day_index]
worst_week=1 if worst_day_index < 7 else 2

# print results
print(f"Total sales: KES{total_sales}")
print(f"Average Daily Sales: KES{average_sales:,.2f}")
print(f"Highest sales: KES{highest_sale}(Day:{best_day_name},Week:{best_week})")
print(f"Lowest Sale: KES {lowest_sale} (Day:{worst_day_name}'Week:{worst_week})")


```

## Task 2:Week totals and Growth
* My apprach:I wil written a function that uses slicing '[:7]' to grab week 1 and '[7:]' to grab week 2.Then calculate the percentage growth from week 1 to week 2.*

```python
def week_totals(sales):
  """Return a tuple of (week 1 total,week 2 total)."""
  week1 = sum(sales[:7])
  week2 = sum(sales[7:])
  return(week1, week2)

#Unpack the tuple
w1_total, w2_total = week_totals(daily_sales)

# Calculate growth percentage
growth = ((w2_total - w1_total)/w1_total )*100

print(f"Week 1 Total: KES{w1_total}")
print(f"Week 2 Total: KES{w2_total}")
print(f"Week-on Week Growth: {growth:.2f}%")
```

#Restock Analysis and Gross Margin
*My approach:I will write a function that loops through the stock dictionary.I will check if the quantity is less than or equal to the reorder level.I will also calculate the gross margin for each item.*


```python
def get_restock_items(stock_dict):
  """ Return a list of product that need restocking."""
  restock_list = []
  for product, details in stock_dict.items():
    if details["qty"] <= details["reorder"]:
      restock_list.append(product)
  return restock_list

items_to_restock = get_restock_items(stock)
print(f"Items that need restocking: {items_to_restock}")

print("\nGross Margin per product:")
for product, details in stock.items():
  margin= ((details["price"]-details["cost"])/details["price"]) * 100
  print(f"{product}: {margin:.2f}%")

```
